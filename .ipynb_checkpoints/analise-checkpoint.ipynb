```python
%matplotlib inline

import pandas as pd 
import matplotlib.pyplot as plt

dados = pd.read_csv("dados/alunos.csv", sep=";")
```

```python
idades = dados["idade"].sort_values()
idades.describe()
```

```python
dados["idade"].hist(bins=5)

plt.title("Distribuição das idades")
plt.xlabel("Idade")
plt.ylabel("Quantidade de alunos")
plt.show()
```
